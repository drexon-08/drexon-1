```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
data["diagnosis"]=le.fit_transform(data["diagnosis"])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=data.drop("diagnosis",axis=1)
y=data["diagnosis"]
```

```python
x.shape
```

```python
y.head
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
```

```python
#logistic regression
from sklearn.linear_model import LogisticRegression
```

```python
model_lr=LogisticRegression(max_iter=5000)
model_lr.fit(x_train_scaled,y_train)
```

```python
y_pred_lr=model_lr.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score,confusion_matrix
```

```python
print("WITHOUT PCA")
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall:", recall_score(y_test, y_pred_lr))
print("F1 Score:", f1_score(y_test, y_pred_lr))
```

```python
confusion_matrix(y_test,y_pred_lr)
```

```python
X_boundary = x[["texture_mean", "area_mean"]]
X_boundary_train, X_boundary_test, y_boundary_train, y_boundary_test = (train_test_split(X_boundary,y,test_size=0.20,random_state=42))
```

```python
X_boundary.shape
```

```python
boundary_scaler = StandardScaler()
X_boundary_train_scaled = boundary_scaler.fit_transform(X_boundary_train)

X_boundary_test_scaled = boundary_scaler.transform(X_boundary_test)
```

```python
boundary_model = LogisticRegression(max_iter=5000)
boundary_model.fit(X_boundary_train_scaled,y_boundary_train)
```

```python
from sklearn.inspection import DecisionBoundaryDisplay
import seaborn as sns
```

```python
y_boundry_train_labels = y_train.map({0: "Benign",1: "Malignant"})
```

```python
DecisionBoundaryDisplay.from_estimator(boundary_model,X_boundary_train_scaled,response_method="predict",alpha=0.3)

sns.scatterplot(x=X_boundary_train_scaled[:, 0],y=X_boundary_train_scaled[:, 1],hue=y_boundry_train_labels)
plt.xlabel("Texture Mean (Scaled)")
plt.ylabel("Area mean (Scaled)")
plt.title("Decision Boundary Without PCA")

plt.show()
```

```python
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2)
```

```python
x_train_pca=pca.fit_transform(x_train_scaled)
x_test_pca=pca=pca.fit_transform(x_test_scaled)
```

```python
x_train_pca.shape
```

```python
x_test_pca.shape
```

```python
print("Features before PCA:", x_train_scaled.shape)
print("Features after PCA:", x_train_pca.shape)
```

```python
model_with_pca = LogisticRegression(max_iter=5000)
model_with_pca.fit(x_train_pca, y_train)
```

```python
pred_with_pca=model_with_pca.predict(x_test_pca)
```

```python
print("WITH PCA")
print("Accuracy:", accuracy_score(y_test, pred_with_pca))
print("Precision:", precision_score(y_test, pred_with_pca))
print("Recall:", recall_score(y_test, pred_with_pca))
print("F1 Score:", f1_score(y_test, pred_with_pca))
```

```python
confusion_matrix(y_test,pred_with_pca)
```

```python
pca.explained_variance_
```

```python
DecisionBoundaryDisplay.from_estimator(model_with_pca,x_train_pca,response_method="predict",alpha=0.5)

sns.scatterplot(x=x_train_pca[:, 0],y=x_train_pca[:, 1],hue=y_train_labels,edgecolor="black")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Logistic Regression Decision Boundary With PCA")
plt.show()
```
