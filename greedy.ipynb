```python

print("*** Enter Graph ***")
graph = {}
heu={}

n = int(input("Enter Number of Nodes: "))
for i in range(n):
    node = input(f"\nEnter Node {i+1}: ")
    graph[node] = []
        
    value=int(input("Enter heuristic value of node: "))
    heu[node]=value
        

    neighbours = int(input(f"Enter Number of Neighbours of {node}: "))
    for j in range(neighbours):
        neighbour = input(f"Enter Neighbour {j+1}: ")
        graph[node].append(neighbour)

print("\nGraph:")
for node in graph:
    print(node, ":", graph[node])
        
print("\nHeu: ") 
for node in heu:
    print(node,":",heu[node])


# graph={
#     'a':['b','c','d'],
#     'b':['a','e'],
#     'c':['a','f','e'],
#     'd':['a','f'],
#     'f':['c','d','g'],
#     'e':['b','c','h'],
#     'g':['f','h'],
#     'h':['e','h']
# }

# heu={
#     'a':40,
#     'b':32,
#     'c':25,
#     'd':35,
#     'f':17,
#     'e':19,
#     'g':0,
#     'h':10
# }

# graph={
#     'p':['r','c','a'],
#     'r':['p','e','c'],
#     'c':['p','r','v','m'],
#     'e':['r','v','s'],
#     'v':['c','e','s','m','n'],
#     'a':['p','m'],
#     'm':['c','v','l'],
#     'l':['l','n'],
#     'n':['l','v','s'],
#     's':['e','v','n']
# }

# heu={
#     'p':10,
#     'a':11,
#     'r':8,
#     'c':6,
#     'e':3,
#     'v':4,
#     'm':9,
#     'l':9,
#     'n':6,
#     's':0
# }
        
        
def greedy(graph,heu,start,goal):
    openlist=[start]
    closelist=[]
    print('intial openlist: ',openlist)
    print('intial closelist: ',closelist)
    print()
    
    while len(openlist)!=0:#until openlist has element
        current=openlist[0]
        
        for node in openlist:
            if heu[node]<heu[current]:
                current=node
                
        openlist.remove(current)
        closelist.append(current)
        print("======================")
        print('current:',current)
        print('openlist: ',openlist)

        
        #goal
        if current== goal:
            print("goal found")
            print("path: ",closelist)
            return
        
        # neighbors 
        for neighbor in graph[current]:
            if(neighbor in closelist):
                continue
            if (neighbor not in openlist):
                current=neighbor
                openlist.append(current)
            print("added: ",current)
            print("openlist: ",openlist)
            print('closelist:',closelist)
            print("------------------------")
                
             
            
    
        
## main 
print("==== greedy best first search ====")
start=input("Enter start value: ")
goal=input("Enter goal value: ")
greedy(graph,heu,start,goal)
```
